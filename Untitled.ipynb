```python
import numpy as np
import plotly.graph_objects as go
from ipywidgets import widgets, Output
from IPython.display import display

# 1. 積分対象の関数 (正則関数)
def f(z):
    return z**2

# 2. 積分計算ロジック
def compute_integral(path_complex):
    if len(path_complex) < 2:
        return 0j
    # 各区間の中点での関数値 * 区間の幅 (長方形近似)
    z_mid = (path_complex[:-1] + path_complex[1:]) / 2
    dz = np.diff(path_complex)
    return np.sum(f(z_mid) * dz)

# 3. UIとグラフの設定
fig = go.FigureWidget()
fig.add_scatter(x=[], y=[], mode='lines+markers', name='Path')
fig.update_layout(
    title="複素積分シミュレーター: 軌跡を描いてください",
    xaxis=dict(range=[-2, 2], title="Re(z)"),
    yaxis=dict(range=[-2, 2], title="Im(z)"),
    width=600, height=600,
    template="plotly_white",
    dragmode='drawopenpath' # 自由曲線の描画モード
)

out = widgets.Output()

def update_integral(change):
    # 最新の描画データを取得
    shapes = fig.layout.shapes
    if not shapes:
        return
    
    last_shape = shapes[-1]
    x = last_shape['xcoords'] if 'xcoords' in last_shape else [] # Plotlyのバージョンによる
    # 手動取得が不安定な場合があるため、pathデータから抽出
    path_str = last_shape.get('path', "")
    if not path_str: return

    # パス文字列を座標リストに変換
    try:
        coords = [p.replace('M', '').replace('L', '').split(',') for p in path_str.split() if p not in ['M', 'L', 'Z']]
        path_complex = np.array([complex(float(c[0]), float(c[1])) for c in coords])
        
        integral_val = compute_integral(path_complex)
        
        with out:
            out.clear_output()
            print(f"軌跡の点数: {len(path_complex)}")
            print(f"積分値 ∫ f(z) dz = {integral_val:.4f}")
            if abs(path_complex[0] - path_complex[-1]) < 0.2:
                print("💡 ほぼ閉曲線です。積分値が 0 に近づいているか確認してください！")
    except:
        pass

fig.observe(update_integral, names='layout')

display(widgets.VBox([fig, out]))
```
