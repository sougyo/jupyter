```julia
using Plots

# パラメータと媒介変数の定義
a = 1.0          # 半径
t = 0:0.01:2π    # t の範囲（0 から 2πまで）

# 媒介変数の計算
x = a * cos.(t)
y = a * sin.(t)

# グラフの描画
plot(x, y, label="x = a cos(t), y = a sin(t)", aspect_ratio=:equal)
xlabel!("x")
ylabel!("y")
title!("Parametric Plot: Circle")
```

```julia
using Plots

# x の範囲を定義
x = -3:0.01:3  # -3から3まで0.01刻み

# cosh(x) の計算
y = cosh.(x) .* cosh.(x)

# グラフの描画
plot(x, y, 
    label = "cosh^2(x)", 
    xlabel = "x", 
    ylabel = "cosh(x)", 
    title = "Graph of cosh(x)", 
    color = :blue, 
    linewidth = 2, 
    legend = :topright
)
```
