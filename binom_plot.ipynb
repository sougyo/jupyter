```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import binom

# パラメータの組み合わせ（n, p）
param_sets = [
    (10, 0.1),
    (10, 0.5),
    (10, 0.9),
    (20, 0.5),
    (50, 0.5)
]

x_values = [np.arange(0, n+1) for n, _ in param_sets]

# グラフの描画
fig, axes = plt.subplots(nrows=2, ncols=3, figsize=(15, 8))
axes = axes.flatten()

for i, (n, p) in enumerate(param_sets):
    x = x_values[i]
    y = binom.pmf(x, n, p)
    axes[i].bar(x, y, color='skyblue', edgecolor='black')
    axes[i].set_title(f'n={n}, p={p}')
    axes[i].set_xlabel('k')
    axes[i].set_ylabel('P(X=k)')

# 最後のサブプロットは使わない
fig.delaxes(axes[-1])

fig.suptitle('PMF', fontsize=16)
plt.tight_layout()
plt.show()
```
