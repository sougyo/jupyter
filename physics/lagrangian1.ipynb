```julia
using Symbolics
```

```julia
@variables t
```

```julia
@variables x(t)[1:2] y(t)[1:2]
```

```julia
@variables L(x[1], x[2], y[1], y[2], t)
```

```julia
@variables m g
```

```julia
L = (1//2) * m * (y[1]^2 + y[2]^2) - m * g * x[2]
```

```julia
eq = Dict()
for i in 1:2
    eq[i] = expand_derivatives(Differential(t)(Differential(y[i])(L))) ~ expand_derivatives(Differential(x[i])(L))
end
```
