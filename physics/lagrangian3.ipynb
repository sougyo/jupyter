```julia
using Symbolics
```

```julia
@variables x[1:4]
```

```julia
@variables E(x[1], x[2], x[3], x[4])[1:3], B(x[1], x[2], x[3], x[4])[1:3]
```

```julia
@variables F(E[1], E[2], E[3], B[1], B[2], B[3])[1:4, 1:4]
```

```julia
eq = substitute(F[1,1]^2, Dict(F[1,1] => E[1]))
display(expand_derivatives(Differential(x[1])(eq)))
```
