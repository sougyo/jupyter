```julia
using Symbolics
```

```julia
@variables t
```

```julia
@variables x(t)[1:3] v(t)[1:3]
```

```julia
@variables L(x[1], x[2], x[3], v[1], v[2], v[3], t)
```

```julia
@variables c m
```

```julia
DIM=2
```

```julia
L = - m * c^2 * sqrt(1 - (sum(v[i]^2 for i in 1:DIM))/(c^2))
```

```julia
eq = Dict()
for i in 1:DIM
    eq[i] = simplify(expand(expand_derivatives(Differential(t)(Differential(v[i])(L)) - Differential(x[i])(L))))
end
```

```julia
for i in 1:DIM
    display(eq[i])
end
```
